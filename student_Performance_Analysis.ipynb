```python
import matplotlib.pyplot as plt
```

```python
grade_counts = df["Final Grade"].value_counts().sort_index()

plt.figure(figsize=(6,4))
grade_counts.plot(kind="bar")
plt.title("Grade Distribution")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["Study Hours"], df["Exam Score"])
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df["Attendance (%)"], df["Exam Score"])
plt.title("Attendance vs Exam Score")
plt.xlabel("Attendance (%)")
plt.ylabel("Exam Score")
plt.show()
```

```python
gender_avg = df.groupby("Gender")["Exam Score"].mean()

plt.figure(figsize=(5,4))
gender_avg.plot(kind="bar")
plt.title("Gender-wise Average Exam Score")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.show()
```

```python
top_students = df.sort_values(by="Exam Score", ascending=False).head(10)

plt.figure(figsize=(8,4))
plt.bar(top_students["Student ID"], top_students["Exam Score"])
plt.title("Top 10 Students by Exam Score")
plt.xlabel("Student ID")
plt.ylabel("Exam Score")
plt.xticks(rotation=45)
plt.show()
```

## Student Performance Insights - Students with higher study hours generally achieved better exam scores. - Better attendance is associated with improved academic performance. - Most students scored in the middle-to-high grade range. - The top-performing students combined good attendance with consistent study habits. - Regular assignments and study time contribute to stronger overall performance.
